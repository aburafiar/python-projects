w (write)is to open a new file in  write mode IF FILE ALREADY EXIST IT OVERWRITE THE THE OLD FILE DATA

W+(write and then read) same as write but first opens in write mode then in read mode IF FILE ALREADY EXIST IT OVERWRITE THE THE OLD FILE DATA


a open new file in write mode If file already exists then it opens file write mode then the data is appended to the existing file
