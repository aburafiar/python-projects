```python
lines_to_write = [
    "First line of text\n",
    "Second line of text\n",
    "Third line of text\n",
    "Fourth line of text\n"
]

file_name = "/content/sample_data/my_lines.txt"

with open(file_name, "w") as fp:
  fp.writelines(lines_to_write)

print(f"Content written to {file_name}:\n")
with open(file_name, "r") as fr:
  print(fr.read())
```

```python
sample_lines = [
    "This is the first line.\n",
    "This is the second line.\n",
    "And this is the third line.\n",
    "Finally, the fourth line."
]

read_example_file = "/content/sample_data/read_example.txt"

# First, write some content to the file for demonstration
with open(read_example_file, "w") as f_write:
  f_write.writelines(sample_lines)

print("--- Using read() ---")
# Use read() to read the entire file content as a single string
with open(read_example_file, "r") as f_read:
  full_content = f_read.read()
  print("Full content:\n", full_content)

print("\n--- Using readlines() ---")
# Use readlines() to read all lines into a list of strings
with open(read_example_file, "r") as f_readlines:
  list_of_lines = f_readlines.readlines()
  print("List of lines:\n", list_of_lines)
```

```python
file_for_readline = "/content/sample_data/readline_example.txt"
lines_for_readline = [
    "Line 1: This is the first line.\n",
    "Line 2: This is the second line.\n",
    "Line 3: This is the third line.\n",
    "Line 4: This is the fourth line."
]

# First, write some content to the file
with open(file_for_readline, "w") as f_write:
  f_write.writelines(lines_for_readline)

print("--- Using readline() ---")
with open(file_for_readline, "r") as f_read_one_by_one:
  # Read the first line
  first_line = f_read_one_by_one.readline()
  print(f"First call to readline(): {first_line.strip()}") # .strip() removes leading/trailing whitespace, including newline

  # Read the second line
  second_line = f_read_one_by_one.readline()
  print(f"Second call to readline(): {second_line.strip()}")

  # You can loop through the rest of the lines
  print("\nLooping through remaining lines:")
  for line in f_read_one_by_one:
    print(line.strip())

  # After all lines are read, readline() returns an empty string
  empty_line = f_read_one_by_one.readline()
  print(f"\nAfter reading all lines, readline() returns: '{empty_line}' (an empty string)")
```
